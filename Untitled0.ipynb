```python
# Step 2A: Load dataset
with open("data/raw/Pride_and_Prejudice-Jane_Austen.txt", "r", encoding="utf-8") as f:
    text = f.read()

print("Characters in dataset:", len(text))
print(text[:500])  # show sample
```

```python
import re

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s,.!?;:()\'"-]', ' ', text)   # keep basic characters
    text = re.sub(r'\s+', ' ', text).strip()              # clean spaces
    return text

clean_text = preprocess_text(text)

print("Cleaned text length:", len(clean_text))
print(clean_text[:500])
```

```python
tokens = clean_text.split()
print("Total tokens:", len(tokens))
print(tokens[:50])
```

```python
from collections import Counter

counter = Counter(tokens)
vocab = sorted(counter.keys())

itos = {i:word for i, word in enumerate(vocab)}  # index → word
stoi = {word:i for i, word in itos.items()}      # word → index

print("Vocabulary size:", len(vocab))
```

```python
numerical_tokens = [stoi[token] for token in tokens]
print("Example numerical tokens:", numerical_tokens[:50])
```

```python
seq_len = 40
```

```python
seq_len = 40
```

```python
inputs = []
targets = []

for i in range(len(numerical_tokens) - seq_len):
    inp = numerical_tokens[i : i + seq_len]
    tgt = numerical_tokens[i + 1 : i + seq_len + 1]
    inputs.append(inp)
    targets.append(tgt)

print("Total sequences created:", len(inputs))
```

```python
dataset = list(zip(inputs, targets))

import random
random.shuffle(dataset)
```

```python
N = len(dataset)
train_data = dataset[: int(0.90 * N)]
val_data   = dataset[int(0.90 * N) : int(0.95 * N)]
test_data  = dataset[int(0.95 * N) : ]

print("Train samples:", len(train_data))
print("Val samples:", len(val_data))
print("Test samples:", len(test_data))
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

class LMDataset(Dataset):
    def __init__(self, data_path):
        self.data = torch.load(data_path)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        inp, tgt = self.data[idx]
        return torch.tensor(inp), torch.tensor(tgt)
```

```python
def get_loaders(batch_size=64):
    train_ds = LMDataset("data/processed/train.pt")
    val_ds   = LMDataset("data/processed/val.pt")
    test_ds  = LMDataset("data/processed/test.pt")

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
    val_loader   = DataLoader(val_ds, batch_size=batch_size, shuffle=False)
    test_loader  = DataLoader(test_ds, batch_size=batch_size, shuffle=False)

    return train_loader, val_loader, test_loader
```

```python
import json

with open("data/processed/itos.json", "r") as f:
    itos = json.load(f)

vocab_size = len(itos)
print("Vocabulary size:", vocab_size)
```

```python
import torch.nn as nn

class LSTMLanguageModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=200, hidden_dim=400, num_layers=2, dropout=0.3):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, embed_dim)

        self.lstm = nn.LSTM(
            embed_dim,
            hidden_dim,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True
        )

        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x, hidden=None):
        x = self.embedding(x)
        output, hidden = self.lstm(x, hidden)
        logits = self.fc(output)
        return logits, hidden
```

```python
model = LSTMLanguageModel(vocab_size)
print(model)
```

```python
import torch
import torch.nn as nn
from torch.optim import Adam
import math

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Move model to device
model = model.to(device)

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = Adam(model.parameters(), lr=0.001)

# Load data
train_loader, val_loader, test_loader = get_loaders()
```

```python
def evaluate(loader):
    model.eval()
    total_loss = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            logits, _ = model(x)
            loss = criterion(logits.view(-1, vocab_size), y.view(-1))
            total_loss += loss.item()

    return total_loss / len(loader)
```

```python
best_val = float('inf')
epochs = 3  # you can increase later

for epoch in range(1, epochs + 1):
    model.train()
    total_loss = 0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)

        optimizer.zero_grad()
        logits, _ = model(x)
        loss = criterion(logits.view(-1, vocab_size), y.view(-1))
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_train = total_loss / len(train_loader)
    avg_val = evaluate(val_loader)
    ppl = math.exp(avg_val)

    print(f"Epoch {epoch} | Train Loss: {avg_train:.4f} | Val Loss: {avg_val:.4f} | Perplexity: {ppl:.2f}")

    # Save best model
    if avg_val < best_val:
        best_val = avg_val
        torch.save(model.state_dict(), "best_model.pt")
        print("✅ Saved best model!")
```

```python
test_loss = evaluate(test_loader)
print("Test Loss:", test_loss)
print("Perplexity:", math.exp(test_loss))
```

```python
import torch

def generate_text(model, start_text, max_new_tokens=50):
    model.eval()

    # split words
    words = start_text.split()

    # convert to ids
    input_ids = torch.tensor([[stoi[w] for w in words]], device=device)

    hidden = None

    for _ in range(max_new_tokens):
        logits, hidden = model(input_ids, hidden)

        # get last token's logits
        last_logits = logits[0, -1]

        # pick most likely next word
        next_id = torch.argmax(last_logits).item()

        # add word to output
        words.append(itos[str(next_id)])

        # prepare next input
        input_ids = torch.tensor([[next_id]], device=device)

    return " ".join(words)
```

```python
prompt = "elizabeth was"
print(generate_text(model, prompt, max_new_tokens=50))
```
